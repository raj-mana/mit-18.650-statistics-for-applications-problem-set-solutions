# Problem 1 $\:$ Convergence of Random Variables

1.  Suppose that $\P(X_n = \frac 1n) = 1-\frac{1}{n^2}$ and $\P(X_n = n) = \frac{1}{n^2}$. Then $X_n \to 0$ in probability.

    If $X_n$ converges in $L^2$, then it converges to $0$ in $L^2$. However, $\E|X_n|^2 \ge n^2 \cdot \frac{1}{n^2} = 1$. So $X_n$ does not converge in $L^2$.

2.  Let $X_n \sim \Ber(p)$ be IID.

    (a)  $n \bar X_n \sim \text{Binomial}(n,p).$

    (b) By the law of large numbers, $\bar X_n \to p$ almost surely. So by the bounded convergence theorem, $\bar X_n \to p$ in $L^2.$

3.  Let $X_n \sim \Pois(\frac 1n)$.

    (a) A direct computation shows that $X_n \to 0$ in probability.

    (b) Similarly, $nX_n \to 0$ in probability.

# Problem 2 $\:$ True or False

1.  If $X_n \to X$ and $Y_n \to Y$ almost surely, then $\P(X_n \to X \text{ and }Y_n\to Y) = 1$, so $X_n + Y_n \to X+Y$ almost surely.

2.  If $X_n \to X$ and $Y_n \to Y$ in probability, we show that $X_n + Y_n \to X + Y$ in probability.

    Suppose that $\varepsilon > 0$. Let $\delta > 0$. There exists $N$ such that $n \ge N$ implies $\P(|X_n - X| \le \varepsilon/2) \ge 1-\delta$ and $\P(|Y_n - Y| \le \varepsilon/2) \ge 1-\delta.$ Hence

    \begin{align*}
    \P(|(X_n-Y_n) - (X-Y)| > \varepsilon) &\le
    \P(|X_n-X| > \varepsilon/2 \text{ or } |Y_n - Y| > \varepsilon/2)\\
    &\le  \P(|X_n-X| > \varepsilon/2) + \P(|Y_n-Y| > \varepsilon/2)\\
    &\le 2\delta.
    \end{align*}

Then $X_n \rightsquigarrow X$ and $Y_n \rightsquigarrow X$. But $X_n + Y_n = 0$ does not converge to $X + X \sim N(0,2)$.

# Problem 3 $\:$ A Confidence Interval for Bernoulli Random Variables

Let $X_1,\dots,X_n \sim \Ber(p)$ be IID, where $p \in (0,1)$. Let $Z \sim N(0,1).$

1.  By the central limit theorem, $\sqrt n \frac{\bar X_n - p}{\sqrt{p(1-p)}} \rightsquigarrow Z.$

2.  $\P(|Z| \le t) = 2 \P(Z \le t) - 1$. To remember this, think of the density of $N(0,1)$ and visualize $2 \P(Z \le t) - \P(|Z| \le t) = 1.$

3.  Let $I_t = \bar X_n \pm \frac{t\sqrt{p(1-p)}}{\sqrt n}$. Then
    
    \begin{align*}
    \P(p \in I_t ) \to \P( |Z| \le t ) = 2 \Phi(t) - 1
    \end{align*}
    
    as $n \to \infty.$

4.  Let $t_0 = 1.96.$ Then $\P(p \in I_{t_0} ) \to 0.95$ as $n\to \infty.$

5.  (a) For all $p$, we have $p(1-p) \le \frac 14.$

    (b) Let $J_1 = \bar X_n \pm \frac{1.96}{2\sqrt n}.$ Then $\liminf_{n\to \infty}\P(p\in J_1) \ge 0.95.$

    (c) If $p \le 0.3$, we can use $ \bar X_n \pm \frac{1.96 \sqrt{0.3 \times 0.7}}{\sqrt n}$ in (b).

6.  (a) That $p \in I_{t_0}$ is equivalent to

    $$(\bar X_n - p)^2 \le \frac{t_0^2 p(1-p)}{n}.$$

    (b) We find that
    
    $$\frac{\bar X_n+\frac{t_0^2}{2n}-t_0\sqrt{\frac{\bar X_n(1-\bar X_n)}{n}+\frac{t_0^2}{4n^2}}}{1+\frac{t_0^2}{n}}\le p\le\frac{\bar X_n+\frac{t_0^2}{2n}+t_0\sqrt{\frac{\bar X_n(1-\bar X_n)}{n}+\frac{t_0^2}{4n^2}}}{1+\frac{t_0^2}{n}}.$$


    (c) Let $J_2$ be the interval with endpoints the left-hand side and right-hand side of the inequality in (b). Then $\P(p\in J_2) \to 0.95 $ as $n\to \infty.$

7.  An application of the law of large numbers shows that $\sqrt{\bar X_n (1-X_n)} \to \sqrt{p(1-p)}$ almost surely. Then, by the central limit theorem and Slutzky's theorem,
    
    $$ \frac{\sqrt{n}(\bar X_n - p)}{\sqrt{\bar X_n (1-\bar X_n)}} = \frac{\sqrt{n}(\bar X_n - p)}{\sqrt{p (1-p)}} \frac{\sqrt{p(1-p)}}{\sqrt{\bar X_n(1-\bar X_n)}}\rightsquigarrow N(0,1). $$

    So

    $$ J_3 = \bar X_n \pm \frac{1.96 \sqrt{\bar X_n(1-\bar X_n)}}{\sqrt n}$$

    contains $p$ with probability $\to 0.95.$

8.  $n$ voters are sample independently and with replacement. Let $p$ denote the proportion of voters who plan to vote for Hillary and $\hat p$ the proportion of sampled voters who plan to vote for Hillary.

    (a) Let $J_1,J_2,J_3$ be as in the previous problems. Suppose that $n=10,000$ and $\hat p = 0.7341.$

```python
import numpy as np

n = 10_000
hat_p = 0.7341
t_0 = 1.96

J_1 = hat_p - t_0 / (2 * np.sqrt(n)) , hat_p + t_0 / (2 * np.sqrt(n))
J_2 = (hat_p + t_0**2 / (2*n) - t_0 * np.sqrt(hat_p*(1-hat_p)/n + t_0**2 / (4*n**2))) / (1 + t_0**2 / n) , (hat_p + t_0**2 / (2*n) + t_0 * np.sqrt(hat_p*(1-hat_p)/n + t_0**2 / (4*n**2))) / (1 + t_0**2 / n)
J_3 = hat_p - t_0 * np.sqrt(hat_p * (1-hat_p) / n) , hat_p + t_0 * np.sqrt(hat_p * (1-hat_p) / n)

print(J_1, J_2, J_3)

l = lambda x : x[1]-x[0]

print(l(J_1), l(J_2), l(J_3))
```
