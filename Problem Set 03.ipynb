# Problem 1 $\:$ Maximum Likelihood Estimator

Let $X_1,\dots, X_n \sim f_\theta$ be IID  with $\theta \in (0,\infty)$ and let $\hat \theta$ denote the MLE of $\theta.$

1. $f_\theta = \theta \tau^\theta x^{-(\theta + 1)} \1(x \ge \tau).$ Then
   \begin{align*}
   \ell_n(\theta) &= n \theta \log \tau + n \log \theta - (\theta + 1) \sum_{i = 1}^n \log X_i, \\
   \frac{\partial \ell_n}{\partial \theta} &= n \log \tau + \frac n \theta - \sum_{i=1}^n \log X_i.
   \end{align*}
   So $\lim_{\theta\to 0^+} \ell_n(\theta) = -\infty$, $\lim_{\theta \to \infty} \ell_n(\theta) = -\infty$ (we use $X_i > \tau$ a.s.). Solving $ \frac{\partial \ell_n}{\partial \theta}(\hat \theta) = 0$, we find that
   $$\hat \theta = \left (\frac 1n \left ( \sum_{i=1}^n \log X_i \right )  - \log \tau \right)^{-1}.$$
2. $f_\theta = \tau \theta^\tau x^{-(\tau + 1)} \1(x \ge \theta).$ Then
   \begin{align*}
   \ell_n(\theta) &= n \log \tau + n \tau \log \theta - (\tau + 1) \sum_{i=1}^n \log X_i   
   \end{align*}
   on $(0, X_{(1)}]$. We see that $\ell_n(\theta)$ is strictly increasing on $(0, \min(X_1,\dots,X_n)],$ so
   $$\hat\theta = X_{(1)} .$$
3. $f_\theta = \sqrt{\theta} x^{\sqrt{\theta}-1} \1(0 \le x \le 1).$ Then
   \begin{align*}
   \ell_n(\theta) &= \frac n2 \log \theta + (\sqrt \theta - 1 ) \sum_{i=1}^n \log X_i, \\
   %
   \frac{\partial \ell_n}{\partial \theta} &= \frac{n}{2\theta} + \frac{1}{2\sqrt \theta} \sum_{i=1}^n \log X_i.
   \end{align*}
   So $\lim_{\theta\to 0^+}\ell_n(\theta) = -\infty$, $\lim_{\theta\to\infty} \ell_n(\theta ) = -\infty$ (since $\sum_{i=1}^n \log X_i < 0$ a.s.). Solving $ \frac{\partial \ell_n}{\partial \theta}(\hat \theta) = 0$, we find that
   $$ \theta = \left ( \frac 1n \left ( \sum_{i=1}^n \log X_i\right) \right)^{-2}. $$
4. $f_\theta = \frac{x}{\theta^2} \exp\left(-\frac{x^2}{2\theta^2}\right) \1(x \ge 0).$ Then
   \begin{align*}
   \ell_n(\theta) &= \sum_{i=1}^n \log X_i - 2n \log \theta - \frac{1}{2\theta^2} \sum_{i=1}^n X_i^2,\\
   \frac{\partial \ell_n}{\partial \theta} &= -\frac{2n}{\theta} + \frac{1}{\theta^3} \sum_{i=1}^n X_i^2.
   \end{align*}
   So $\lim_{\theta \to 0^+} \ell_n(\theta) = -\infty$, $\lim_{\theta \to \infty} \ell_n(\theta) = -\infty.$ Solving $ \frac{\partial \ell_n}{\partial \theta}(\hat \theta) = 0$, we find that
   $$ \hat \theta = \sqrt{\frac 1{2n} \sum_{i=1}^n X_i^2}.$$
5. $f_\theta = \theta \tau x^{\tau-1} \exp(-\theta x^\tau) \1(x \ge 0).$ Then
   \begin{align*}
   \ell_n(\theta) &= n \log \theta + n \log \tau + (\tau - 1) \sum_{i=1}^n \log X_i - \theta \sum_{i=1}^n X_i^\tau, \\
   \frac{\partial \ell_n}{\partial \theta} &= \frac n\theta - \sum_{i=1}^n X_i^\tau.
   \end{align*}
   So $\lim_{\theta \to 0^+} \ell_n(\theta) = -\infty$, $\lim_{\theta \to \infty} \ell_n(\theta) = -\infty.$ Solving $ \frac{\partial \ell_n}{\partial \theta}(\hat \theta) = 0$, we find that
   $$ \hat\theta = \left ( \frac 1n \sum_{i=1}^n X_i^\tau\right )^{-1}. $$
   

# Problem 2 $\:$ Consistency of the Maximum Likelihood Estimator

Let $X_1,\dots,X_n \sim N(\theta,\theta)$ be IID. Then
\begin{align*}
\ell_n(\theta) &= - \frac n2 \log (2\pi) - \frac n2 \log \theta - \frac{1}{2\theta} \sum_{i=1}^n (X_i - \theta)^2, \\
\frac{\partial\ell_n}{\partial \theta} &= -\frac n{2\theta} + \frac 1{2\theta^2} \sum_{i=1}^n (X_i - \theta)^2 + \frac 1 \theta \sum_{i=1}^n (X_i - \theta) \\
&= -\frac n{2\theta}  + \frac{1}{2\theta^2} \sum_{i=1}^n X_i^2 - \frac n2.
\end{align*}
Observe that $\lim_{\theta \to 0^+} \ell_n(\theta) = -\infty$ a.s. and $\lim_{\theta \to \infty} \ell_n(\theta) = -\infty$. Solving $\frac{\partial\ell_n}{\partial \theta}(\hat \theta_{\text{MLE}}) = 0$,  
$$ \hat \theta_{\text{MLE}} = \frac{-1 + \sqrt{1 + \frac 4n \sum_{i=1}^n X_i^2}}{2}. $$
As $n\to \infty$, the law of large numbers states that $\frac 1n \sum_{i=1}^n X_i^2 \to \E X_1^2 = \var X_1 + (\E X_1)^2 = \theta + \theta^2$ a.s., so a.s.
\begin{align*}
\hat \theta_{\text{MLE}} &\to \frac{-1 + \sqrt{1 + 4( \theta + \theta^2)}}{2} = \theta.
\end{align*}

# Problem 3 $\:$ Kullback–Leibler Divergence
The Kullback–Leibler Divergence between $\P_\theta$ and $\P_{\theta'}$ is defined to be
$$ \newcommand{KL}{\operatorname{KL}} \KL(\P_\theta, \P_{\theta'}) = \E_\theta \log \left ( \frac{f_{\theta}}{f_{\theta'}}\right ),$$
where $f_{-}$ is a PDF for continuous measures and a PMF for discrete measures ($\P_\theta$ and $\P_{\theta'}$ are both discrete or both continuous).
1. Let $\E$ denote expectation with respect to $N(a,\sigma^2)$. We compute
\begin{align*}
\KL(N(a,\sigma^2), N(b,\sigma^2)) &=  \E \left( \frac{(X-b)^2 - (X-a)^2}{2\sigma^2} \right) \\
&= \E \left( \frac{-\sigma^2 + \E X^2 - 2b \E X + b^2}{2\sigma^2} \right) \\
&= \E \left( \frac{-\sigma^2 + \sigma^2 + a^2 - 2ab + b^2}{2\sigma^2} \right) \\
&=  \frac{(a-b)^2}{2\sigma^2} .
\end{align*}
2. Let $0 < a,b < 1$. We compute
\begin{align*}
\KL(\Ber(a), \Ber(b)) &= \log\left(\frac{1-a}{1-b}\right)(1-a) + \log\left( \frac ab \right) a.
\end{align*}

# Problem 4 $\:$ Total Variation Distance
1. $\newcommand{\TV}{\operatorname{TV}}$ Let $0 < s \le t.$ We have
\begin{align*}
\TV(\Uni([0,s]), \Uni([0,t])) &= \frac 12\int_0^t \left |\frac{1}{s}\1(x \le s) - \frac 1t \right| dx \\
&= \frac{1}{2} s \left ( \frac 1s - \frac 1t\right) + \frac 12 (t-s) \frac 1t \\
&= 1 - \frac st.
\end{align*}
2. Let $0 \le p \le q \le 1$ Then
\begin{align*}
\TV(\Ber(p), \Ber(q)) &= \frac 12 \left ( (1-p) - (1-q) + q-p \right) \\
&= q-p.
\end{align*}
3. Let $X_1, \dots, X_n \sim \Ber(p)$ with $p \in [0,1].$ Then, by (2),
\begin{align*}
\TV(\Ber(\bar X_n), \Ber(p)) &=  |\bar X_n - p|.
\end{align*}
By the law of large numbers, $\bar X_n \to p$ almost surely, so $\TV(\Ber(\bar X_n), \Ber(p)) \to 0$ almost surely.
4. We have
\begin{align*}
\TV\left (\Pois\left (\frac 1n\right ), \delta_0\right)
&= \frac 12 (1 - e^{-1/n}) + \frac 12 (1 - e^{-1/n}) \to 0
\end{align*}
as $n\to \infty.$
