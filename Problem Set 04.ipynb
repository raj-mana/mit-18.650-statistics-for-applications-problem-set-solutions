# Problem 1 $\: $ Maximum Likelihood and Fisher Information
Let $X_1, \dots, X_n$ be IID.

1. $X_1 \sim \Ber(p)$ for $p\in(0,1).$ Then
   \begin{align*}
   \ell_n(p) &= \sum_{i=1}^n ((1-X_i)\log(1-p) + X_i\log(p)), \\
   \frac{\partial\ell_n}{\partial p} &= \sum_{i=1}^n \left (\frac{X_i}{p} - \frac{1-X_i}{1-p} \right).
   \end{align*}
   Since $ \lim_{p \to 0^+}\ell_n(p) = \lim_{p\to 1^-} \ell_n(p) = -\infty$, we see that $\hat p_{\text{MLE}}$ is the solution to $\frac{\partial\ell_n}{\partial p} (\hat p_{\text{MLE}}) = 0.$ Hence
   $$\hat p_{\text{MLE}} = \bar X_n. $$

   The Fisher information is equal to
   \begin{align*}
   I(p) &= -\E_p\left ( \frac{\partial^2 \ell_1}{\partial p^2}\right) \\
   &= \E_p \left ( \frac{X_1}{p^2} + \frac{1-X_1}{(1-p)^2}\right ) \\
   &= \frac 1{p(1-p)}.
   \end{align*}

2. $X_1 \sim \Pois(\lambda)$ with $\lambda > 0.$ Then
   \begin{align*}
   \ell_n(\lambda) &= -n\lambda + (\log \lambda) \sum_{i=1}^n X_i - \sum_{i=1}^n \log X_i!, \\
   \frac{\partial \ell_n}{\partial \lambda} &= -n + \frac{1}{\lambda} \sum_{i=1}^n X_i.
   \end{align*}
   Since $ \lim_{\lambda \to 0^+}\ell_n(\lambda) = \lim_{\lambda\to \infty} \ell_n(\lambda) = -\infty$, we see that  $\hat \lambda_{\text{MLE}}$ is the solution to $\frac{\partial\ell_n}{\partial \lambda} (\hat \lambda_{\text{MLE}}) = 0.$ So
   $$\hat \lambda_{\text{MLE}} = \bar X_n. $$
   The Fisher information is equal to
   \begin{align*}
   I(\lambda) = - \E_\lambda \left ( \frac{\partial^2 \ell_1}{\partial \lambda^2}\right) = \frac 1\lambda.
   \end{align*}

3. $X_1 \sim \Exp(1/\lambda)$ with $\lambda > 0.$ Then
   \begin{align*}
   \ell_n(\lambda) &= n \log \lambda - \lambda \sum_{i=1}^n X_i, \\
   \frac{\partial \ell_n(\lambda)}{\partial \lambda} &= \frac n\lambda - \sum_{i=1}^n X_i.
   \end{align*}
   Since $ \lim_{\lambda \to 0^+}\ell_n(\lambda) = \lim_{\lambda\to \infty} \ell_n(\lambda) = -\infty$, we see that  $\hat \lambda_{\text{MLE}}$ is the solution to $\frac{\partial\ell_n}{\partial \lambda} (\hat \lambda_{\text{MLE}}) = 0.$ So
   $$\hat \lambda_{\text{MLE}} = \frac{1}{\bar X_n}. $$
   The Fisher information is equal to
   $$ I(\lambda) = - \E_\lambda \left ( \frac{\partial^2 \ell_1}{\partial \lambda^2}\right) = \frac 1 {\lambda^2}.$$

4. $X_1 \sim N(\mu, \sigma^2)$ with $\mu \in \R$ and $\Sigma = \sigma^2 > 0.$ Then
   \begin{align*}
   \ell_n(\mu, \Sigma) &= -\frac n2 \log(2\pi) - \frac n2\log( \Sigma) - \frac 1{2\Sigma} \sum_{i=1}^n (X_i - \mu)^2, \\
   \frac{\partial \ell_n}{\partial \mu} &=  \frac {1}{\Sigma} \sum_{i=1}^n (X_i - \mu) , \\
   \frac{\partial \ell_n}{\partial \Sigma} &= -\frac{n}{2\Sigma} + \frac{1}{2\Sigma^2} \sum_{i=1}^n (X_i-\mu)^2.
   \end{align*}
   Observe that, to maximize $\ell_n(\mu, \Sigma)$, we can first find $\hat \mu_{\text{MLE}}$ that maximizes $-\sum_{i=1}^n (X_i-\mu)^2$, that is,
   $$ \hat \mu_{\text{MLE}} = \bar X_n. $$
   Then $\lim_{\Sigma \to 0^+} \ell_n(\hat \mu_{\text{MLE}}, \Sigma) =  \lim_{\Sigma \to \infty} \ell_n(\hat \mu_{\text{MLE}}, \Sigma) = -\infty$ implies $\frac{\partial \ell_n}{\partial \Sigma}(\hat \mu_{\text{MLE}}, \hat\Sigma_{\text{MLE}}) = 0$, so that
   $$ \hat \Sigma_{\text{MLE}} = \frac 1n \sum_{i=1}^n (X_i - \bar X_n)^2.$$

   Now,
   \begin{align*}
   \frac{\partial^2 \ell_1}{\partial \mu^2} &= -\frac{1}{\Sigma}, \\
   \frac{\partial^2 \ell_1}{\partial \mu\partial \Sigma} = \frac{\partial^2 \ell_1}{\partial \Sigma\partial \mu} &= -\frac{X_1 - \mu}{\Sigma^2}, \\
   \frac{\partial^2 \ell_1}{\partial \Sigma^2} &=  \frac{1}{2\Sigma^2} - \frac{1}{\Sigma^3} (X_1 - \mu)^2,
   \end{align*}
   So
   $$ I(\mu, \Sigma) = \begin{bmatrix} \frac 1 \Sigma & 0 \\ 0 & \frac{1}{2\Sigma^2}\end{bmatrix}. $$

5. $X_1 \sim f_{a,\lambda}$ with $f_{a,\lambda}(x) = \lambda e^{-\lambda(x-a)} \1(x\ge a)$ for $a\in \R$, $\lambda >0$, $x\in\R.$ Then
   \begin{align*}
   \ell_n(a, \lambda) &= n \log \lambda - \lambda \sum_{i=1}^n (X_i - a).
   \end{align*}
   for $a \in (-\infty, X_{(1)}]$, $\lambda >0$. To maximize $\ell_n(a, \lambda),$ for fixed $\lambda$ we can maximize in $a$ to find that
   $$ \hat a_{\text{MLE}} = X_{(1)}.$$
   We see that $\lim_{\lambda \to 0^+} \ell_n(\hat a_{\text{MLE}}, \lambda) = \lim_{\lambda \to \infty} \ell_n(\hat a_{\text{MLE}}, \lambda) = -\infty$, so $\frac{\partial \ell_n}{\partial \lambda} (\hat a_{\text{MLE}}, \hat \lambda_{\text{MLE}}) = 0$. Hence
   $$ \hat \lambda_{\text{MLE}} = \frac{n}{\sum_{i=1}^n (X_i - X_{(1)}) }. $$

   Now,
   \begin{align*}
   \frac{\partial\ell_1}{\partial a} &= \lambda, \\
   \frac{\partial\ell_1}{\partial \lambda} &= \frac 1\lambda- (X_1 - a) = \frac 1 \lambda + a - X_1.
   \end{align*}
   Also,
   \begin{align*}
   \E(X_1) &= \int_a^\infty x \lambda e^{-\lambda(x-a)} dx \\
   &= \int_0^\infty (t+a) \lambda e^{-\lambda t } dt \\
   &= \frac 1\lambda + a, \\
   \E(X_1)^2 &= \frac 1{\lambda^2} + 2a \frac 1 \lambda + a^2,\\
   \E(X_1^2) &= \int_a^\infty x^2 \lambda e^{-\lambda(x-a)} dx \\
   &= \int_0^\infty (t+a)^2 \lambda e^{-\lambda t} dt \\
   &= \int_0^\infty t^2 \lambda e^{-\lambda t} dt + 2a \int_0^\infty t \lambda e^{-\lambda t} dt + a^2 \int_0^\infty \lambda e^{-\lambda t} dt \\
   &= \frac 1{\lambda^2} + \frac 1{\lambda^2} + 2a \frac{1}{\lambda} + a^2, \\
   \var X_1 &= \frac 1{\lambda^2}.
   \end{align*}
   The covariance of the score function is
   $$ \E(D\ell_1) (D\ell_1)^T = \begin{bmatrix}0 & 0\\ 0& \frac 1 {\lambda^2}\end{bmatrix}. $$
   However,
   $$ \frac{\partial \ell_1}{\partial a \partial \lambda} = 1,$$
   so the Fisher information is not well-defined.

6. $X_1 \sim f_{\mu, \sigma^2}$ with $f_{\mu, \sigma^2}(x) = \frac{1}{x\sqrt{2\pi \sigma^2}} e^{-\frac{1}{2\sigma^2}(\log x - \mu)^2}$ for $\mu\in\R$, $\Sigma = \sigma^2 > 0$, $x > 0.$ Then
   \begin{align*}
   \ell_n(\mu, \Sigma) &= - \sum_{i=1}^n \log X_i - \frac{n}{2} \log 2\pi - \frac{n}{2} \log \Sigma - \frac{1}{2\Sigma} \sum_{i=1}^n (\log X_i - \mu)^2, \\
   \frac{\partial\ell_n}{\partial \mu} &= \frac 1\Sigma \sum_{i=1}^n (\log X_i - \mu), \\
   \frac{\partial\ell_n}{\partial \Sigma} &= - \frac n {2\Sigma} + \frac{1}{2\Sigma^2} \sum_{i=1}^n (\log X_i - \mu)^2.
   \end{align*}
   To maximize $\ell_n(\mu, \Sigma)$, for fixed $\Sigma$ we can maximize in $\mu$ to find that
   $$ \hat \mu_{\text{MLE}} = \frac 1n \sum_{i=1}^n \log X_i.$$
   Since $\lim_{\Sigma \to 0^+} \ell_n(\hat \mu_{\text{MLE}}, \Sigma) =
   \lim_{\Sigma \to \infty} \ell_n(\hat \mu_{\text{MLE}}, \Sigma) = -\infty$, we see that $\frac{\partial\ell_n}{\partial \Sigma}(\hat \mu_{\text{MLE}}, \hat \Sigma_{\text{MLE}}) = 0.$ Hence
   $$ \hat \Sigma_{\text{MLE}} = \frac 1n \sum_{i=1}^n (\log X_i - \hat \mu_{\text{MLE}})^2. $$
   Now,
   \begin{align*}
   \frac{\partial^2 \ell_1}{\partial \mu^2} &= -\frac 1 \Sigma, \\
   \frac{\partial^2 \ell_1}{\partial \mu \partial \Sigma} =
   \frac{\partial^2 \ell_1}{\partial \Sigma \partial \mu} &= - \frac 1 {\Sigma^2} (\log X_1 - \mu), \\
   \frac{\partial^2 \ell_1}{\partial \Sigma^2} &= \frac{1}{2\Sigma^2} - \frac{1}{\Sigma^3}(\log X_1 - \mu)^2.
   \end{align*}
   We compute the distribution of $\log X_1.$ We see that
   \begin{align*}
   F(t) &= \P(\log X_1 \le t) \\ &= \P(X_1 \le e^t) \\
   &= \int_{0}^{e^t} \frac{1}{x\sqrt{2\pi \sigma^2}} e^{-\frac{1}{2\sigma^2}(\log x - \mu)^2} \, dx \\
   F'(t) &= \frac{1}{e^t \sqrt{2\pi \sigma^2}} e^{-\frac{1}{2\sigma^2}(t - \mu)^2} \cdot e^t \\
   &= \frac{1}{ \sqrt{2\pi \sigma^2}} e^{-\frac{1}{2\sigma^2}(t - \mu)^2},
   \end{align*}
   and so $\log X_1 \sim N(\mu, \sigma^2).$ Hence
   $$ I(\mu, \Sigma) = \begin{bmatrix} \frac 1 \Sigma & 0\\ 0& \frac{1}{2\Sigma^2}\end{bmatrix}.$$

# Problem 2 $\: $ Method of Moments
For each of the distributions in Problem 1, we compute the method of moments estimator. Let $X_1, \dots, X_n$ be IID.

1. $X_1 \sim \Ber(p).$ Then
   $$ \hat p = \hat m_1 = \bar X_n. $$
2. $X_1 \sim \Pois(\lambda).$ Then
   $$ \hat \lambda = \hat m_1 = \bar X_n. $$
3. $X_1 \sim \Exp(1/\lambda).$ Then
   $$ \hat \lambda = \frac{1}{\hat m_1} =  \frac{1}{\bar X_n}. $$
4. $X_1 \sim N(\mu, \sigma^2).$ (Recall that the second parameter is $\sigma^2.$) Then
   \begin{align*}
   \hat \mu &= \hat m_1 = \bar X_n
   \end{align*}
   and
   \begin{align*}
   \hat \sigma^2 + \hat \mu^2 &= \hat m_2 = \frac{1}{n} \sum_{i=1}^n X_i^2, \\
   \hat \sigma^2 &= \frac{1}{n} \sum_{i=1}^n X_i^2 - (\bar X_n)^2 = \frac{1}{n}\sum_{i=1}^n (X_i - \bar X_n)^2.
   \end{align*}
5. $X_1 \sim f_{a,\lambda}$ with $f_{a,\lambda}(x) = \lambda e^{-\lambda(x-a)} \1(x\ge a)$ for $a\in \R$, $\lambda >0$, $x\in\R.$ By the computations in Problem 1,
   \begin{align*}
   \hat m_1 &= \frac{1}{\hat\lambda} + \hat a \\
   \hat m_2 - \hat m_1^2 &= \frac 1{\hat\lambda^2}.
   \end{align*}
   Thus,
   \begin{align*}
   \hat \lambda &= (\hat m_2 - \hat m_1^2)^{-1/2}, \\
   \hat a &= \hat m_1 - \hat \lambda^{-1}.
   \end{align*}
6. $X_1 \sim f_{\mu, \sigma^2}$ with $f_{\mu, \sigma^2}(x) = \frac{1}{x\sqrt{2\pi \sigma^2}} e^{-\frac{1}{2\sigma^2}(\log x - \mu)^2}$ for $\mu\in\R$, $\Sigma = \sigma^2 > 0$, $x > 0.$ Also note that $\sigma^2$ is the parameter, so the hat in $\hat \sigma^2$ should really be over $\sigma^2$. By the computation in Problem 1, $\log X_1 \sim N(\mu, \sigma^2).$ If $X \sim N(0,1),$ then $\E (e^{tX}) = e^{t^2 / 2}.$ So
   \begin{align*}
   \E( X_1^t ) &= \E(e^{t \log X_1} )\\
   &= \E(e^{t (\mu + \sigma X)} )\\
   &= e^{t\mu} e^{t^2 \sigma^2 / 2}, \\
   \E(X_1) &= e^{\mu + \sigma^2/2}, \\
   \E(X_1^2) &= e^{2(\mu + \sigma^2)}.
   \end{align*}
   Then
   \begin{align*}
   \hat\mu + \frac 12 \hat \sigma^2  &= \log \hat m_1, \\
   2 \hat \mu + 2 \hat \sigma^2 &= \log \hat m_2,
   \end{align*}
   which means that
   \begin{align*}
   \hat\mu &= 2\log \hat m_1 - \frac{1}{2}\log \hat m_2,\\
   \hat\sigma^2 &= \log \hat m_2 - 2\log \hat m_1.
   \end{align*}

# Problem 3 $ \: $ Censored Data
In a given population, $n$ individuals are sampled with replacement to ask if their salaries exceed $z$ (fixed). Assume that the population distribution of salaries follow $\Exp(1/\lambda).$ Denote by $X_1,\dots,X_n \in \{0,1\}$ the binary responses of the sampled individuals and $Y_1, \dots, Y_n$ the salaries of the sampled individuals.

1. $X_1 \sim \Ber(\P(\Exp(1/\lambda) > z)) = \Ber(e^{-\lambda z}).$

2. By the central limit theorem, $\bar X_n$ is asympotically normal with asympotic variance $e^{-\lambda z}(1-e^{-\lambda z}).$

3. By the law of large numbers, $\bar X_n \to e^{-\lambda z}$ a.s., so $-\frac 1z \log \bar X_n \to \lambda$ a.s.

4. By the delta method,
$$ \sqrt n\left ( -\frac 1z \log(\bar X_n) - \lambda\right ) \approx N\left(0, \frac{e^{2\lambda z}}{z^2} e^{-\lambda z}(1-e^{-\lambda z})\right) =
N\left(0, \frac{e^{\lambda z}-1}{z^2}  \right).$$

5. Consider $h(z) = \frac{e^{\lambda z}-1}{z^2}$ for $z > 0.$ Then $\lim_{z\to 0^+} h(z) = \lim_{z\to\infty} h(z) = \infty.$ If $z$ minimizes $h$, then $h'(z) = 0$, so that
   $$ \frac{\lambda z e^{\lambda z} - 2(e^{\lambda z} - 1)}{z^3} = 0$$
   Thus,
   $$z = \frac{2}{\lambda}(1-e^{-\lambda z}). $$
   This is a fixed-point formulation for $z.$

6. (a) The statistical model for $Y_1,\dots Y_n$ is $\Exp(1/\lambda)$ for $\lambda >0.$
   
   (b) We compute the Fisher information for the statistical model in (a). We have
   \begin{align*}
   \ell_1(\lambda) &= \log \lambda - \lambda Y_1 \\
   \ell_1'(\lambda) &= \frac 1 \lambda - Y_1 \\
   \ell_1''(\lambda) &= - \frac 1 {\lambda^2},
   \end{align*}
   so
   $$I_Y(\lambda) = \frac 1 {\lambda^2}. $$
   
   (c) In the model where only the $X$ are observed, in which $z$ is fixed,
   \begin{align*}
   \ell_1(\lambda) &= (1-X_1)\log (1- e^{-\lambda z})  + X_1 \log e^{-\lambda z} \\
   &= (1-X_1)\log (1- e^{-\lambda z}) - X_1 \lambda z,\\
   \ell_1'(\lambda) &= (1-X_1)\frac{z}{e^{\lambda z} - 1} - X_1 z,\\
   \ell_1''(\lambda) &= -(1-X_1) \frac{z^2 e^{\lambda z}}{(e^{\lambda z }-1)^2},
   \end{align*}
   so
   $$ I_X(\lambda) = \frac{z^2}{e^{\lambda z} - 1}.$$
   
   (d) We notice that
   \begin{align*}
   \frac{z^2}{e^{\lambda z} - 1} < \frac{1}{\lambda^2}
   \end{align*}
   if and only if
   \begin{align*}
   \frac 1 {e^{\lambda z} - 1} \le \frac 1{(\lambda z)^2},
   \end{align*}
   if and only if
   $$ x^2 < e^x - 1$$
   for all $x > 0.$ But it is true that $x^2 < e^x - 1$ for all $x > 0.$ Hence $I_X(\lambda) < I_Y(\lambda).$ Heuristically, this means more information is obtained from observing $Y_1,\dots,Y_n$ than from observing $X_1,\dots,X_n.$
