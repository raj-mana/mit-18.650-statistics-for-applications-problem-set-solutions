# Problem 1 $\:$ Biased/Unbiased Estimation

Let $X_1,\dots,X_n \sim \Ber(p)$ be IID with $p\in(0,1).$

1.  $\var X_1 = p(1-p).$

2.  By applying the law of large numbers, $\bar X_n (1-\bar X_n) \to p(1-p)$ almost surely, hence in probability, so $\bar X_n(1-\bar X_n)$ is a consistent estimator of $p(1-p)$.

3.  We have
    \begin{align*}
    \E \bar X_n (1-\bar X_n) &= \E \bar X_n - \E \bar X_n^2 \\
    &= p - \var \bar X_n - (\E \bar X_n)^2 \\
    &= p - \frac{p(1-p)}{n} - p^2 \\
    &= p(1-p) \left ( 1 - \frac 1n\right).
    \end{align*}
    So

    $$ \bias \bar X_n(1-\bar X_n) = -\frac{p(1-p)}{n}. $$

4.  Using (3), an unbiased estimator of $p(1-p)$ is

    $$ \frac{\bar X_n(1-\bar X_n)}{1 - \frac 1n}. $$

# Problem 2 $\:$ Statistical Models and Identifiability

We define a *statistical model* to be a pair $(E,\{\P_\theta\}_{\theta \in \Theta})$ where $E$ is a set and $\{\P_\theta\}_{\theta \in \Theta}$ is a family of probability distributions on $E$ parameterized by $\Theta.$ A statistical model is *identifiable* if $\theta \mapsto \P_\theta$ is injective.

1.  We observe $n$ IID $\Pois(\lambda)$'s. The statistical model is $(\mathbf Z_{\ge 0}, \{\Pois(\lambda)\}_{\lambda \in (0,\infty)} )$ and is identifiable.
2.  We observe $n$ IID $\Exp(1/\lambda)$'s with $\lambda \le 10.$ The statistical model is $((0,\infty), \{\Exp(1/\lambda)\}_{0 < \lambda \le 10})$ and is identifiable.
3.  We observe $n$ IID $\Uni([0,\theta])$'s. The statistical model is $([0,\infty), \{\Uni([0,\theta])\}_{\theta > 0}) $ and is identifiable.
4.  We observe $n$ IID $N(\mu, \sigma^2)$'s. The statistical model is $(\R, \{N(\mu,\sigma^2))\}_{\mu \in \R, \sigma > 0}) $ and is identifiable.
5.  We observe the sign of $n$ IID $N(\mu, \sigma^2)$'s. Let $\P_{\mu, \sigma}(1) = \P( N(0,1) > - \frac \mu \sigma) = 1 - \Phi(-\frac \mu\sigma)$ and $\P_{\mu, \sigma}(-1) = 1- \P_{\mu, \sigma}(1)$. The statistical model is $(\{\pm 1 \}, \{\P_{\mu,\sigma}\}_{\mu \in \R, \sigma >0} )$ and is not identifiable.
6.  We use a statistical procedure to observe $n$ IID $\Uni([0,\theta])$'s. The statistical model is $([0,\infty), \{\Uni([0,\theta])\}_{\theta > 0}) $ and is identifiable.
7.  We sample $n$ individuals, with replacement, and ask if their commute time is at least $20$ minutes. The commute time of a random person is assumed to follow an exponential distribution.

    The statistical model is $ ( \\{0,1 \\} , \{\Ber(e^{-20\lambda}) \}_{\lambda > 0}) $ and is identifiable.
8.  We sample the lifespan of $67$ machines, capping at $500$ days. The lifespan of each machine follows an exponential distribution.

    Let $X(\lambda) \sim \Exp(1/\lambda)$ and let $\P_\lambda$ be the distribution of $\min(X(\lambda), 500).$ The statistical model is $\\{(0,500) , \\{\P_\lambda \\}_{\lambda > 0 } \\}$ and is identifiable (one can condition one the lifespan being $<500$ to see this).

# Problem 3 $\:$ A Confidence Interval for Poisson Distributions

Let $X_1,\dots, X_n \sim \Pois(\lambda)$ be IID. Let $Z \sim N(0,1).$

1. $\sqrt{\frac{n}{\lambda}}(\bar X_n - \lambda) \rightsquigarrow Z.$

2. $\P(|Z| \le t) = 2 \P(Z \le t) - 1.$

3. Let $I = \bar X_n \pm 1.96 \sqrt{\frac{\lambda}{n}}.$ Then $\P(\lambda \in I ) \to 0.95$ as $n\to \infty.$

4. Let $J = \bar X_n \pm 1.96 \sqrt{\frac{\bar X_n}{n}}.$ Then $\P(\lambda \in J ) \to 0.95$ as $n\to \infty.$

# Problem 4 $\:$ A Confidence Interval for Uniform Distributions

Let $X_1,\dots,X_n \sim \Uni([0,\theta])$ be IID. Define $M_n = \max_{1\le i \le n} X_i.$

1. Let $\varepsilon > 0.$ Then (assume $\varepsilon < \theta$)
\begin{align*}
\P(|M_n - \theta| > \varepsilon) &=
\P(X_i < \theta - \varepsilon \text{ for }1\le i \le n) \\
&= \left (\frac{\theta-\varepsilon}{\theta} \right)^n \\
&\to 0
\end{align*}
as $n\to\infty.$

2. We compute
\begin{align*}
\P\left(n\left(1-\frac{M_n}{\theta}\right) \le t\right) &=
\P\left( M_n \ge \theta \left (  1 - \frac tn\right)\right ) \\
&= 1 - \left (  1 - \frac tn\right)^n
\end{align*}
for $t\in[0,n].$ As $n\to \infty,$ we see that the CDF of $n\left(1-\frac{M_n}{\theta}\right)$ converges pointwise to $1-e^{-t}$ for $t\in(0,\infty),$ which is the CDF of a random variable $\sim \Exp(1).$ So $n\left(1-\frac{M_n}{\theta}\right) \rightsquigarrow \Exp(1).$
3. If $ n\left(1-\frac{M_n}{\theta}\right) \le t$, then $\theta \le  \frac{M_n}{1-t/n}$ for $n>t$. In particular, choose $t$ so that $\P(\Exp(1) \le t) = 0.95$, that is, $t = - \log 0.05.$ Then
\begin{align*}
\P\left (M_n \le \theta \le  \frac{M_n}{1-t/n}\right)
&= \P (n(1 - M_n/\theta) \le t) \\
&\to \P(\Exp(1) \le t) \\
&= 0.95,
\end{align*}
so if $I = [M_n, \frac{M_n}{1-t/n}],$ then $\P(\theta \in I) \to 0.95$ as $n\to \infty.$

4. The expectation of $M_n$ is $\theta(1- \frac{1}{n+1})$, so $M_n$ is not unbiased.
